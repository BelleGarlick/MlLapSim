```python
import os
import sys
import os
from toolkit.maths import segment_intersections
import numpy as np
import matplotlib.pyplot as plt

sys.path.append(os.path.abspath("src"))

data_path = "/Users/belle/Downloads/Real Test"
simulations = os.listdir(data_path)


def open_path(f):
    with open(f, "r") as file:
        return file.read()

def open_xyrl_file(f):
    data = open_path(f).split("\n")[1:]
    data = [x.split(",") for x in data if len(x) > 1]
    data = [[float(b) for b in x] for x in data]
    return data


import toolkit.tracks.conversion

def check_collision(file_data):
    track = toolkit.tracks.conversion.from_xyrl(data=file_data)

    seg_arrs = np.array([seg.arr() for seg in track.segmentations])
    all_intersections = []
    for idx, seg in enumerate(track.segmentations):
        pre_lines = segment_intersections(seg_arrs[idx], seg_arrs[:idx])
        post_lines = segment_intersections(seg_arrs[idx], seg_arrs[idx+1:])
        all_intersections += pre_lines + post_lines
    return len(all_intersections) > 0

def check_simulation(simulation):
    boundary_file = open_xyrl_file(data_path + "/" + simulation + "/" + "boundaries.csv")
    track_file = open_xyrl_file(data_path + "/" + simulation + "/" + "track.csv")

    return check_collision(track_file) or check_collision(boundary_file)


import multiprocessing

with multiprocessing.Pool() as pool:
    output = pool.map(check_simulation, simulations)
```
