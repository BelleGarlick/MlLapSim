```python
# 2. Define the Model
class SimpleCNN(nn.Module):
    def __init__(self, n_kernel=8):
        super(SimpleCNN, self).__init__()
        # in_channels=1, out_channels=1, kernel_size=5
        # self.conv = nn.Conv1d(in_channels=1, out_channels=n_kernel, kernel_size=5)
        self.fc1 = nn.Linear(10, n_kernel)
        self.fc2 = nn.Linear(n_kernel, 1)
        # self.fc3 = nn.Linear(2, 1)
        self.act = nn.Sigmoid()

    def forward(self, x):
        # x = self.conv(x)
        x = x.view(x.size(0), -1) # Flatten
        x = self.act(self.fc1(x))
        x = self.act(self.fc2(x))
        # x = self.act(self.fc3(x))
        return x

model = SimpleCNN()
criterion = nn.BCELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# 3. Simple Training Loop
for epoch in range(1000):
    optimizer.zero_grad()
    outputs = torch.max(model(x), model(-x))
    loss = criterion(outputs, yt)
    loss.backward()
    optimizer.step()

print(f"Final Loss: {loss.item():.4f}")
```
